```python
# 0/1 Knapsack using Dynamic Programming

def knapsack(weights, values, capacity):
    n = len(weights)

    # Create DP table
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

    # Fill the DP table
    for i in range(1, n + 1):   #iterate throgh each item
        for w in range(1, capacity + 1):      #checks every possible capacity

            if weights[i - 1] <= w:     #current item can fit into the current capacity
                dp[i][w] = max(
                    values[i - 1] + dp[i - 1][w - weights[i - 1]],    #value of current item best value possible with the remaining capacity
                    dp[i - 1][w]
                )
            else:
                dp[i][w] = dp[i - 1][w]   #we copy the result from the previous row

    return dp[n][capacity]    #returns the maximum value


# Example
weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 5

max_value = knapsack(weights, values, capacity)

print("Maximum value:", max_value)
```
