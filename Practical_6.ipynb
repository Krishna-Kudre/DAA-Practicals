```python
# Chain Matrix Multiplication using Dynamic Programming

def matrix_chain_order(p):
    n = len(p) - 1

    # Create DP table
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # length is the chain length
    for length in range(2, n + 1):

        for i in range(n - length + 1):
            j = i + length - 1

            dp[i][j] = float('inf')

            # Try different positions to split the chain
            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i] * p[k + 1] * p[j + 1]
                )

                if cost < dp[i][j]:
                    dp[i][j] = cost

    return dp[0][n - 1]


# Example
# A1 = 10 x 20
# A2 = 20 x 30
# A3 = 30 x 40

dimensions = [10, 20, 30, 40]

minimum_cost = matrix_chain_order(dimensions)

print("Minimum number of scalar multiplications:", minimum_cost)
```
